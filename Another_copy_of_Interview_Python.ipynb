 Examples:

Output : ('l', 2)

Input : geeksforgeeks

Output : ('e', 4)

```python
s = input()
# Create a dictionary to store the frequency of each character
freq = {}

# Count frequency of each character
for char in s:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1

max_char = None
max_count = 0

for char, count in freq.items():
    if count > max_count:
        max_count = count
        max_char = char

print((max_char, max_count))
```

For example,
for N = 4 and S = {1,2,3}, there are four solutions: {1,1,1,1},{1,1,2},{2,2},{1,3}. So output should be 4.

For N = 10 and S = {2, 5, 3, 6}, there are five solutions: {2,2,2,2,2}, {2,2,3,3}, {2,2,6}, {2,3,5} and {5,5}. So the output should be 5.

note: S is the valid coins  50 paise, one, two, five, ten, and twenty rupees for example : In India S ={ 0.5,1,2,5,10,20 }

```python
coins=[1,2,3]
n=4
# Initialize a table to store the number of ways for each value from 0 to n
dp = [0] * (n + 1)

# There is one way to make change for amount 0 (no coins)
dp[0] = 1

# Iterate through each coin in the set
for coin in coins:
    # Update the table based on the current coin
    for i in range(coin, n + 1):
        dp[i] += dp[i - coin]

print (dp[n])
```

**Q.13. Python Program for Find largest prime factor of a number apart from itself.**

Input: 6

Output: 3

Explanation: Prime factor of 6 are - 2, 3, Largest of them is 3.

Input: 15

Output: 5

Note: Input will be  > 2

```python
n = int(input())
largest_factor = 1

# Loop through all numbers from 2 to n
for i in range(2, n + 1):
    # If i is a divisor of n, keep dividing n by i
    if n % i == 0:
        largest_factor = i
        n //= i

    # Stop if n is reduced to 1 (all prime factors have been found)
    if n == 1:
        break

print (largest_factor)
```

```python
def is_symmetric(str):
  n = len(str)
  if str[:n//2] == str[n//2:]:
    return "The entered string is symmetrical"
  else:
    return "The entered string is not symmetrical"

def is_palindrome(str):
  if str == str[::-1]:
      return "The entered string is palindrome"
  else:
    return "The entered string is not palindrome"

inp1 = 'khokho'
inp2 = 'amaama'
print("Input:", inp1)
print(is_symmetric(inp1))
print(is_palindrome(inp1))
print()
print("Input:", inp2)
print(is_symmetric(inp2))
print(is_palindrome(inp2))
```

```python
def calculate_upper_lower(sentence):
    # Initialize counters
    num_upper = 0
    num_lower = 0

    # Iterate through each character in the sentence
    for char in sentence:
        if char.isupper():
            num_upper += 1
        elif char.islower():
            num_lower += 1

    return num_upper, num_lower

# Example input sentence
input_sentence = "Hello world!"

# Calculate and print the number of uppercase and lowercase letters
upper, lower = calculate_upper_lower(input_sentence)
print("UPPER CASE", upper)
print("LOWER CASE", lower)
```

```python
def is_monotonic(lst):
    # We check two conditions:
    # 1. Is it always increasing (or staying the same)?
    # 2. Is it always decreasing (or staying the same)?

    increasing = True
    decreasing = True

    for i in range(len(lst)-1):
        # If we find an element bigger than the current one, it's not increasing
        if lst[i] > lst[i+1]:
            increasing = False
        # If we find an element smaller than the current one, it's not decreasing
        elif lst[i] < lst[i+1]:
            decreasing = False

    # If either 'increasing' or 'decreasing' is still True, the array is monotonic
    return increasing or decreasing

# Let's test it with your examples
print(f"Is [1, 2, 2, 3] monotonic? {is_monotonic([1, 2, 2, 3])}")
print(f"Is [3, 2, 1] monotonic? {is_monotonic([3, 2, 1])}")
print(f"Is [1, 3, 2, 4] monotonic? {is_monotonic([1, 3, 2, 4])}")
```

### 🏫 Lesson: The 'Skipping' Logic
Notice the `if i != j:` line. This is the heart of the problem. It tells Python: 'Multiply everything, but if you are currently looking at the same spot as the outer loop, just skip it.'

**Exercise for you:** Try changing the numbers in `my_numbers` to include a `0` (e.g., `[1, 2, 0, 4]`). Run the code and see what happens to the output!

```python
def product_except_self_beginner(nums):
    answer = []

    # The outer loop: pick one number to 'exclude'
    for i in range(len(nums)):
        current_product = 1

        # The inner loop: multiply all other numbers
        for j in range(len(nums)):
            # If it's not the current number we are skipping...
            if i != j:
                current_product *= nums[j]

        # Add the final result for this spot to our answer list
        answer.append(current_product)

    return answer

# Test it out!
my_numbers = [1, 2, 3, 4]
print(f"Input: {my_numbers}")
print(f"Output: {product_except_self_beginner(my_numbers)}")
```

```python
def two_sum_pointers(numbers, target):
    # Start one pointer at the beginning and one at the end
    left = 0
    right = len(numbers) - 1

    # Continue looping as long as the left pointer is before the right pointer
    while left < right:
        # Calculate the sum of the numbers at the current left and right pointer positions
        current_sum = numbers[left] + numbers[right]

        # If the current sum equals the target, we've found our pair
        if current_sum == target:
            # Return the two numbers that sum up to the target
            return [numbers[left], numbers[right]]

        # If the sum is too small, move the left pointer forward to get a bigger number
        if current_sum < target:
            left += 1
        # If the sum is too big, move the right pointer backward to get a smaller number
        else:
            right -= 1

    # If the loop finishes and no pair is found, return a message
    return "No pair found"

# Let's test it
sorted_data = [1, 2, 4, 6, 8, 10]
my_target = 10

# Call the function with our test data and target
result = two_sum_pointers(sorted_data, my_target)
# Print the target number
print(f"Target: {my_target}")
# Print the result found by the function
print(f"Found pair: {result}")
```

### 🎯 Lesson: Two Pointers Basics
Imagine you have a sorted list of numbers and you want to find two numbers that sum up to 9.
Instead of checking every pair, we use a `left` pointer and a `right` pointer.

```python
def two_sum_two_pointer(nums, target):
    # Initialize pointers at both ends of the sorted array
    left = 0
    right = len(nums) - 1

    while left < right:
        current_sum = nums[left] + nums[right]

        # Scenario 1: Target found
        if current_sum == target:
            return [left, right]  # Returns 0-based indices

        # Scenario 2: Sum is too small -> move left pointer rightward to increase sum
        elif current_sum < target:
            left += 1

        # Scenario 3: Sum is too large -> move right pointer leftward to decrease sum
        else:
            right -= 1

    return []  # Return empty if no pair is found
```

```python
def sorted_squares(nums):
    # Set up two markers: 'left' at the start and 'right' at the end of the list
    left = 0
    right = len(nums) - 1
    # Create an empty list to store our squared results
    result = []

    # Keep looping as long as the markers haven't crossed each other
    while left <= right:
        # Check which number's 'absolute value' (distance from zero) is bigger
        if abs(nums[left]) > abs(nums[right]):
            # If the left one is bigger, square it and add it to our results
            result.append(nums[left] ** 2)
            # Move the left marker one step forward
            left += 1
        else:
            # If the right one is bigger (or equal), square it and add it to results
            result.append(nums[right] ** 2)
            # Move the right marker one step backward
            right -= 1

    # Since we picked the biggest squares first, the list is backwards.
    # This line flips it so it goes from smallest to largest.
    return result#[::-1]
```

```python
def swap_adjacent(lst):
    # We loop through the list, but jump by 2 each time
    # range(start, stop, step)
    for i in range(0, len(lst) - 1, 2):
        # Swap elements at index i and i+1
        lst[i], lst[i+1] = lst[i+1], lst[i]
    return lst

# Example usage:
my_list = [1, 0, 3, 2, 5, 4]
result = swap_adjacent(my_list)

print(f"Original List: [1, 0, 3, 2, 5, 4]")
print(f"Swapped List:  {result}")
```

### Alternative: Swapping using `.append()`
This method is useful when you want to keep your original data unchanged (Immutable approach).

```python
def swap_with_append(lst):
    new_list = []
    # Iterate through the list in steps of 2
    for i in range(0, len(lst) - 1,2):
        # Append the second item first, then the first item
        new_list.append(lst[i+1])
        new_list.append(lst[i])

    # If the list has an odd number of items, add the last one back
    if len(lst) % 2 != 0:
        new_list.append(lst[-1])

    return new_list

# Test
data = [1, 0, 3, 2, 5, 4]
print(f"Original: {data}")
print(f"Swapped (Append method): {swap_with_append(data)}")
```
